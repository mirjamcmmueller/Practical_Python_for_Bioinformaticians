## Indexing

Python starts counting at **zero**. This means it has **0-based indexing**. Indeces are always shown in square brackets `[]`. 
So if I want to know what the first few letters are in the string `sequence`, I can do:

````python
sequence = "ATGCGTAG"

#first letter has index 0
print(sequence[0])
#second letter, index 1
print(sequence[1])
#Third letter, index 2
print(sequence[2])

#Expected output:
#
# A
# T
# G
````

The counting starting at zero might be a bit weird at the beginning, but you get used to it surprisingly quickly.
Eventually your brain will adjust and it will feel normal.

### Negative Indexing
Instead of counting from the start, we can also go backwards i.e. start from the back with negative indeces.

````python
sequence = "ATGCGTAG"

print(sequence[-1])
print(sequence[-2])

#Expected output:
#
# G
# A
````

Using indeces, we can extract part of a string with a range. To denote a range, we use `:`. Extracting a part of a string is called slicing (yes, like cake).

````python
sequence = "ATGCGTAG"

print(sequence[0:4])

#Expected output:
#
# ATGC
````

Note that python does not include the endpoint! Again, a bit weird. But basically 
`[0:4]` means we start **AT** position 0, but stop **BEFORE** position 4. In other words, positions
* 0
* 1
* 2
* 3

are included. Position 4 is excluded.

## Useful String Operations
As Bioinformaticians, we see a lot of strings. ID's, sequences, datafiles and any line we ever read from any file, are strings. 
Obviously I can not show you every single thing you could ever possibly do with a string, but I will show you some common operations that will make your life easier.

First of all, to show the length of a string, use `len()`.

````python
sequence = "ATGCGTAG"

print(len(sequence))

#Expected output:
#
# 8
````

To turn all letters in the string to lowercase, use

````python
print(sequence.lower())
#Expected output:
#
# atgcgtag
````

And if we had a lowercase sequence, this is how we could convert it to uppercase. `.lower()` and `.upper()`are also practical when you want to match strings without being case sensitive or when you want to make a string all lower/upper when it is mixed.

````python
sequence = "atgcgtag"

print(sequence.upper())
#Expected output:
#
# ATGCGTAG
````

# ⚠️ Common beginner mistake

This will NOT work:

```python
number = "5"

print(number + 1)
```

Why?

Because Python cannot add:

* text (`"5"`)
* and numbers (`1`)

You must first convert the string:

```python
number = int(number)

print(number + 1)
```
