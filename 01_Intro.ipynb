## Indexing

Python starts counting at **zero**. This means it has **0-based indexing**. Indeces are always shown in square brackets `[]`. 
So if I want to know what the first few letters are in the string `sequence`, I can do:

```python
sequence = "ATGCGTAG"

#first letter has index 0
print(sequence[0])
#second letter, index 1
print(sequence[1])
#Third letter, index 2
print(sequence[2])

#Expected output:
#
# A
# T
# G
```

The counting starting at zero might be a bit weird at the beginning, but you get used to it surprisingly quickly.
Eventually your brain will adjust and it will feel normal.

### Negative Indexing
Instead of counting from the start, we can also go backwards i.e. start from the back with negative indeces.

```python
sequence = "ATGCGTAG"

print(sequence[-1])
print(sequence[-2])

#Expected output:
#
# G
# A
```

Using indeces, we can extract part of a string with a range. To denote a range, we use `:`. Extracting a part of a string is called slicing (yes, like cake).

```python
sequence = "ATGCGTAG"

print(sequence[0:4])

#Expected output:
#
# ATGC
```

Note that python does not include the endpoint! Again, a bit weird. But basically 
`[0:4]` means we start **AT** position 0, but stop **BEFORE** position 4. In other words, positions
* 0
* 1
* 2
* 3

are included. Position 4 is excluded.

## Useful String Operations
As Bioinformaticians, we see a lot of strings. ID's, sequences, datafiles and any line we ever read from any file, are strings. 
Obviously I can not show you every single thing you could ever possibly do with a string, but I will show you some common operations that will make your life easier.

First of all, to show the length of a string, use `len()`.

```python
sequence = "ATGCGTAG"

print(len(sequence))

#Expected output:
#
# 8
```

To turn all letters in the string to lowercase, use

```python
print(sequence.lower())
#Expected output:
#
# atgcgtag
```

And if we had a lowercase sequence, this is how we could convert it to uppercase. `.lower()` and `.upper()`are also practical when you want to match strings without being case sensitive or when you want to make a string all lower/upper when it is mixed.

```python
sequence = "atgcgtag"

print(sequence.upper())
#Expected output:
#
# ATGCGTAG
```

Replace characters:

```python
sequence = "ATGCGTAG"

print(sequence.replace("A", "X"))

#Expected output:
#
# XTGCGTXG
```

Check if strings starts with something

```python
header = ">seq1"

print(header.startswith(">"))

print(sequence.startswith(">"))

#Expected output:
#
# True
# False

#The equivalent for ending with something is endswith:
print(header.endswith("1"))
#Expected output:
#
# True
```

Joining strings together:

```python
print(sequence + "AAAA")

print("This is the header: " + header)

# Expected output: 
#
# ATGCGTAGAAAA
# This is the header: >seq1
```

```python
gene = "TP53"
count = 15

print(gene, count)

# Expected output:
# TP53 15
```

## Variable type conversion
If it makes sense, we can convert a variable from one type into another. For example:

```python
number = "5"

print(type(number))
```

This is at the moment a string. But we can turn it into an integer:

```python
number = int(number)

print(type(number))
```

Other useful conversions:

```python
#You can turn a string also into a float, a number with decimal.
pi = "3.14"
print(type(float(pi)))
#or to turn a number into a string:
print(type(str(5)))
```

### Beginner mistakes 
This will not work:

```python
number = "5"

print(number + 1)
```

Why?

Because Python cannot add

* text (`"5"`)
* and numbers (`1`)
You can either convert one, or the other. But the outcome will differ:

```python
#If you convert the string to an integer, you can do math with it:
number = int(number)
print(number + 1)

#If you leave number as a string, and add 1 also as a string, you get string combination:
number = "5"
print(number + "1")
```

## Reading error messages
They do look scary yes, and long. But they are usually helpful. For example:


```python
numbers = "123"

print(numbers[10])
```

Important things to look for:

* the error type
* the line number
* the variable involved

The MOST important part is usually near the bottom of the traceback. In this case it tells us that the list index is out of range. Which makes sense, we told it to take the 10th letter, but the string only has 3.

### Common errors and what they mean:

| Error type | What it usually means | Common causes |
|---|---|---|
| `SyntaxError` | Python cannot understand the code you wrote. | Missing `:`, missing brackets, wrong indentation, or using `=` where Python expected something else. |
| `IndentationError` | The spacing at the beginning of lines is inconsistent or incorrect. | Code inside an `if`, `for`, `while`, or function is not indented properly. |
| `NameError` | Python does not know this variable or function name. | Typo in a variable name, or trying to use a variable before creating it. |
| `TypeError` | You are using a value in a way that does not fit its type. | Trying to add a string and an integer, or calling a function with the wrong kind of input. |
| `ValueError` | The type is okay, but the value itself is not valid. | Trying to convert `"hello"` to an integer with `int("hello")`. |
| `IndexError` | You tried to access a position that does not exist. | Using `my_list[10]` when the list has fewer than 11 elements. |
| `KeyError` | You tried to access a dictionary key that does not exist. | Using `my_dict["geneX"]` when `"geneX"` is not in the dictionary. |
| `FileNotFoundError` | Python cannot find the file you asked for. | Wrong file name, wrong folder, or running the notebook from a different location than expected. |
| `AttributeError` | You tried to use a method or attribute that does not exist for that object. | Using a string method on a list, or misspelling a method like `.split()`. |
| `ModuleNotFoundError` | Python cannot find the package you tried to import. | The package is not installed in the current environment/kernel. |
